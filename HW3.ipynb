```python
import numpy as np
import pandas as pd
```

```python
# Series, DataFrame, Index - структуры Pandas

data = pd.Series([0.25, 0.5, 0.75, 1.0])
print(data)
print(type(data))
print(type(data.values))
print(data.index)
print(type(data.index))
```

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0])
print (data[0])
print (data[1:3])
# индексы определяем явно, как хотим
data = pd.Series([0.25, 0.5, 0.75, 1.0], index = ['a', 'b', 'c', 'd'])
print (data)
print (data['a'])
print (data['b':'d'])

print(type(data.index))
```

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0], index = [1, 10, 7, 'd'])
print (data)

print (data[1])
print (data[10:'d'])
```

```python
population_dict = {
    'city_1': 1001,
    'city_2': 1002,
    'city_3': 1003,
    'city_4': 1004,
    'city_5': 1005,
}
population = pd.Series(population_dict)
print (population)

print (population['city_4'])
print (population['city_4':'city_5'])
```

```python
# DataFrame - несколько сирис вместе
population_dict = {
    'city_1': 1001,
    'city_2': 1002,
    'city_3': 1003,
    'city_4': 1004,
    'city_5': 1005,
}
area_dict = {
    'city_1': 1,
    'city_2': 2,
    'city_3': 3,
    'city_4': 4,
    'city_5': 5,
}
population = pd.Series(population_dict)
area = pd.Series(area_dict)

print (area)

states = pd.DataFrame({
    'population1': population,
    'area1': area,
})
print (states)

print (states.values)
print (states.index)
print (states.columns)

print (type(states.values))
print (type(states.index))
print (type(states.columns))

print (states['area1'])
```

```python
# Index
ind = pd.Index([2,3,5,7,11])
print (ind[1])
print (ind[::2])

indA = pd.Index([1,2,3,4,5])
indB = pd.Index([2,3,4,5,6])

print (indA.intersection(indB))
```

```python
# Выборка данных
data = pd.Series([0.25, 0.5, 0.75, 1.0], index = ['a', 'b', 'c', 'd'])
print ('a' in data)
print ('z' in data)

print (data.keys())
print (list(data.items()))

data['a'] = 100
data['z'] = 1000

print (data)

# одномер массив
data = pd.Series([0.25, 0.5, 0.75, 1.0], index = ['a', 'b', 'c', 'd'])

print (data['a':'c'])
print (data[0:2])
print (data[(data > 0.5) & (data < 1)])
print (data[['a', 'd']])
```

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0], index=[1, 3, 10, 15])

print(data[1]) 
print(data.loc[1])
print(data.iloc[1])
```

```python
rng = np.random.default_rng()
s = pd.Series(rng.integers(0, 10, 4))

print(s)
print(np.exp(s))
```

```python
dfA = pd.DataFrame(rng.integers(0, 10, (2, 2)), columns=['a', 'b'])
dfB = pd.DataFrame(rng.integers(0, 10, (3, 3)), columns=['a', 'b', 'c'])

print(dfA)
print(dfB)
print(dfA + dfB)
```

```python
rng = np.random.default_rng(1)
A = rng.integers(0, 10, (3, 4))
print(A)
print(A[0])
print(A - A[0])
```

```python
df = pd.DataFrame(A, columns=['a', 'b', 'c', 'd'])
print(df)
print(df.iloc[0])
print(df - df.iloc[0])
print(df - df.iloc[0, ::2])
```

```python
df = pd.DataFrame(
    [
        [1, 2, 3, np.nan, None, pd.NA],
        [1, 2, 3, None, 5, 6],
        [1, np.nan, 3, None, np.nan, 6]
    ]
)
print(df)
print(df.dropna())
print(df.dropna(axis=0))
print(df.dropna(axis=1))
```

```python
# any , all, thresh
print(df.dropna(axis=1, how="all"))
print(df.dropna(axis=1, how="any"))
print(df.dropna(axis=1, thresh=2))
```
